```python
import os

def find_matching_files(root_dir, filenames):
    matches = []
    filenames_set = set(filenames)  # faster lookup

    for dirpath, _, files in os.walk(root_dir):
        for file in files:
            if file in filenames_set:
                matches.append(os.path.join(dirpath, file))

    return matches


# Example usage
root_directory = "/rs1/researchers/a/amallav/image_dataset"
target_filenames = image_list

results = find_matching_files(root_directory, target_filenames)

for path in results:
    print(path)
```
